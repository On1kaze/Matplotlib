## Задание 1. Построение линейного графика

**Условие:**
Даны значения температуры воздуха за семь дней:
```python
days = [1, 2, 3, 4, 5, 6, 7]
temperature = [12, 15, 14, 17, 19, 18, 21]
```
**Требования:**
- Построить линейный график.
- Добавить маркеры точек.
- Установить синий цвет линии.
- Добавить название графика «Температура воздуха за неделю».
- Подписать оси: «День» и «Температура, °C».
- Включить сетку.


````python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5, 6, 7]
temperature = [12, 15, 14, 17, 19, 18, 21]

plt.figure(figsize=(9, 5))
plt.plot(days, temperature, color="blue", marker="o", markersize=7, linewidth=2, label="Температура")

for d, t in zip(days, temperature):
    plt.annotate(f"{t}°C", (d, t), textcoords="offset points", xytext=(0, 10), ha="center", fontweight="bold")

plt.title("Температура воздуха за неделю", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("День", fontsize=12)
plt.ylabel("Температура, °C", fontsize=12)
plt.xticks(days)
plt.ylim(10, 23)
plt.grid(True, linestyle="--", alpha=0.7)
plt.legend(loc="upper left")
plt.tight_layout()
plt.show()
````

## Задание 2. Сравнение успеваемости

**Условие:**
Даны средние оценки студентов по дисциплинам:
```python
subjects = ['Математика', 'Физика', 'Информатика', 'Английский язык']
grades = [4.1, 3.8, 4.7, 4.3]
```
**Требования:**
- Построить столбчатую диаграмму.
- Использовать разные цвета для столбцов.
- Добавить заголовок и подпись вертикальной оси.
- Определить, по какому предмету средняя оценка самая высокая.


````python
subjects = ['Математика', 'Физика', 'Информатика', 'Английский язык']
grades = [4.1, 3.8, 4.7, 4.3]
colors = ['#3498db', '#e74c3c', '#2ecc71', '#9b59b6']

plt.figure(figsize=(9, 5))
bars = plt.bar(subjects, grades, color=colors, width=0.55, edgecolor="black")

for bar in bars:
    height = bar.get_height()
    plt.annotate(f"{height:.1f}", xy=(bar.get_x() + bar.get_width() / 2, height), xytext=(0, 6), textcoords="offset points", ha="center", va="bottom", fontweight="bold")

plt.title("Средняя успеваемость студентов по дисциплинам", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Дисциплина", fontsize=12)
plt.ylabel("Средняя оценка", fontsize=12)
plt.ylim(0, 5.2)
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()

max_idx = grades.index(max(grades))
print(f"Самая высокая средняя оценка ({grades[max_idx]}) по предмету: «{subjects[max_idx]}».")
````

**Ответ:** Самая высокая средняя оценка (**4.7**) наблюдается по предмету **«Информатика»**.

## Задание 3. Анализ структуры расходов

**Условие:**
Ежемесячные расходы семьи распределены следующим образом:
```python
categories = ['Продукты', 'Транспорт', 'Коммунальные услуги', 'Связь', 'Прочее']
expenses = [120000, 30000, 45000, 15000, 40000]
```
**Требования:**
- Построить круговую диаграмму.
- Отобразить названия категорий.
- Вывести процент каждой категории с помощью параметра `autopct`.
- Добавить заголовок.
- Определить, какая категория занимает наибольшую долю расходов.


````python
categories = ['Продукты', 'Транспорт', 'Коммунальные услуги', 'Связь', 'Прочее']
expenses = [120000, 30000, 45000, 15000, 40000]

total = sum(expenses)
max_idx = expenses.index(max(expenses))
explode = [0.08 if i == max_idx else 0 for i in range(len(categories))]
colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#c2c2f0']

plt.figure(figsize=(8, 7))
plt.pie(expenses, labels=categories, autopct="%1.1f%%", startangle=140, explode=explode, colors=colors, shadow=True, textprops={"fontsize": 11})
plt.title("Анализ структуры ежемесячных расходов семьи", fontsize=14, fontweight="bold", pad=20)
plt.tight_layout()
plt.show()

max_cat = categories[max_idx]
max_val = expenses[max_idx]
share = (max_val / total) * 100
print(f"Наибольшую долю расходов занимает категория «{max_cat}»: {max_val} руб. ({share:.1f}%).")
````

**Ответ:** Наибольшую долю расходов занимает категория **«Продукты»** (120 000 руб., **48.0%** от общих затрат семьи).

## Задание 4. Исследование результатов тестирования

**Условие:**
Даны результаты тестирования студентов:
```python
results = [45, 52, 55, 58, 60, 62, 65, 67, 70, 70, 72, 74, 75, 78, 80, 82, 85, 88, 90, 95]
```
**Требования:**
- Построить гистограмму распределения результатов:
  - количество интервалов — 5;
  - цвет столбцов — оранжевый;
  - цвет границ — чёрный;
  - горизонтальная ось — «Количество баллов»;
  - вертикальная ось — «Количество студентов».
- Сделать вывод: в каком диапазоне находится большинство результатов?


````python
import numpy as np

results = [45, 52, 55, 58, 60, 62, 65, 67, 70, 70, 72, 74, 75, 78, 80, 82, 85, 88, 90, 95]

plt.figure(figsize=(9, 5.5))
counts, bins, patches = plt.hist(results, bins=5, color="orange", edgecolor="black", linewidth=1.2, rwidth=0.95)

for count, patch in zip(counts, patches):
    if count > 0:
        plt.annotate(f"{int(count)} студ.", xy=(patch.get_x() + patch.get_width() / 2, count), xytext=(0, 6), textcoords="offset points", ha="center", va="bottom", fontweight="bold")

plt.title("Распределение результатов тестирования студентов", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Количество баллов", fontsize=12)
plt.ylabel("Количество студентов", fontsize=12)
plt.xticks(bins)
plt.ylim(0, max(counts) + 1.5)
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()

max_bin_idx = int(np.argmax(counts))
print(f"Большинство результатов находится в диапазоне [{bins[max_bin_idx]:.0f} - {bins[max_bin_idx+1]:.0f}] баллов ({int(counts[max_bin_idx])} студентов).")
````

**Вывод:** Большинство результатов (**6 студентов**) находится в диапазоне **[65 - 75] баллов**.

## Задание 5. Исследование связи между показателями

**Условие:**
Дано количество часов подготовки и соответствующие результаты экзамена:
```python
hours = [1, 2, 2, 3, 4, 5, 5, 6, 7, 8]
scores = [48, 52, 55, 60, 65, 70, 73, 78, 84, 90]
```
**Требования:**
- Построить точечный график с помощью функции `scatter()`.
- Ответить на контрольные вопросы:
  1. Какая связь наблюдается между временем подготовки и результатом?
  2. Повышается ли результат при увеличении времени подготовки?
  3. Можно ли на основании этого графика утверждать, что продолжительная подготовка является единственной причиной высокого результата?


````python
hours = [1, 2, 2, 3, 4, 5, 5, 6, 7, 8]
scores = [48, 52, 55, 60, 65, 70, 73, 78, 84, 90]

plt.figure(figsize=(9, 5.5))
plt.scatter(hours, scores, color="#2980b9", s=85, edgecolors="black", linewidth=1.2, zorder=3, label="Результаты студентов")

z = np.polyfit(hours, scores, 1)
p = np.poly1d(z)
x_line = np.linspace(min(hours), max(hours), 100)
plt.plot(x_line, p(x_line), color="#e74c3c", linestyle="--", linewidth=2, label="Линия тренда")

for h, s in zip(hours, scores):
    plt.annotate(f"({h}ч, {s})", (h, s), textcoords="offset points", xytext=(0, 9), ha="center", fontsize=9, alpha=0.85)

plt.title("Зависимость результата экзамена от времени подготовки", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Время подготовки (часы)", fontsize=12)
plt.ylabel("Результат экзамена (баллы)", fontsize=12)
plt.xticks(range(1, 9))
plt.ylim(40, 100)
plt.grid(True, linestyle="--", alpha=0.7)
plt.legend(loc="lower right")
plt.tight_layout()
plt.show()
````

### Ответы на контрольные вопросы:
1. **Какая связь наблюдается между временем подготовки и результатом?**
   > Наблюдается сильная прямая (положительная) линейная связь: чем больше времени студент уделяет подготовке, тем выше его итоговый балл на экзамене.

2. **Повышается ли результат при увеличении времени подготовки?**
   > Да, результат повышается. График демонстрирует явный монотонно возрастающий тренд: при 1 часе подготовки балл равен 48, а при 8 часах достигает 90 баллов.

3. **Можно ли на основании этого графика утверждать, что продолжительная подготовка является единственной причиной высокого результата?**
   > Нет, нельзя. В статистике наличие высокой корреляции между двумя величинами не означает, что одна является *единственной* причиной другой («корреляция не означает причинность»). На итоговую оценку влияют и другие немаловажные факторы: начальный уровень знаний студента, когнитивные способности, качество используемых учебных пособий, методика подготовки, психологическая устойчивость и физическое самочувствие на экзамене.
